```python
from itertools import combinations
from collections import defaultdict

def find_microseismic_centers(lists, threshold):
    # 初始化图结构
    graph = defaultdict(set)
    
    # 遍历所有列表对，构建传递关系
    for i, list1 in enumerate(lists):
        for j, list2 in enumerate(lists):
            if i >= j:  # 避免重复比较
                continue
            for idx1 in list1:
                for idx2 in list2:
                    if abs(idx1 - idx2) <= threshold:
                        # 添加双向边
                        graph[idx1].add(idx2)
                        graph[idx2].add(idx1)
    
    # 寻找连通分量
    visited = set()
    connected_components = []
    
    def dfs(node, component):
        """深度优先搜索，找到连通分量"""
        visited.add(node)
        component.append(node)
        for neighbor in graph[node]:
            if neighbor not in visited:
                dfs(neighbor, component)
    
    for node in graph:
        if node not in visited:
            component = []
            dfs(node, component)
            connected_components.append(component)
    
    # 确定微震中心
    centers = []
    for component in connected_components:
        # 计算每个节点的度数
        degree = {node: len(graph[node]) for node in component}
        # 找到度数最大的节点作为中心
        max_degree = max(degree.values())
        centers_in_component = [node for node, d in degree.items() if d == max_degree]
        centers.append(centers_in_component)
    
    # 返回结果
    return {
        "edges": [(u, v) for u in graph for v in graph[u]],
        "centers": centers
    }

# 示例输入
lists = [
    [1, 5, 10],
    [3, 6, 11],
    [4, 7, 12]
]
threshold = 2

# 调用函数
result = find_microseismic_centers(lists, threshold)

# 输出结果
print("传递序列（边集合）:", result["edges"])
print("可能的微震中心:", result["centers"])
```
